```python
from pyspark.sql import functions as F
from pyspark.sql import types as T

df = spark.table(
    "workspace.student_bronze.raw_student_exams"
)
```

```python
print(f"Number of rows: {df.count():,}")
print(f"Number of columns: {len(df.columns)}")

display(df.limit(10))
```

```python
display(
    df.filter(
        F.col("_rescued_data").isNotNull()
    )
)
```

```python
# 0 czyli git
rescued_count = df.filter(
    F.col("_rescued_data").isNotNull()
).count()

print(f"Rescued records: {rescued_count:,}")
```

```python
# sprawdzenie brakujących wartości
technical_columns = [
    "source_file",
    "loaded_at",
    "_rescued_data"
]

business_columns = [
    c for c in df.columns
    if c not in technical_columns
]

total_rows = df.count()

missing_counts = df.agg(
    *[
        F.count(
            F.when(
                F.col(c).isNull() |
                (F.trim(F.col(c)) == ""),
                1
            )
        ).alias(c)
        for c in business_columns
    ]
).first().asDict()

missing_df = spark.createDataFrame(
    [
        (
            c,
            missing_counts[c],
            round(
                missing_counts[c] / total_rows * 100,
                2
            ) if total_rows else 0.0
        )
        for c in business_columns
    ],
    [
        "column_name",
        "missing_count",
        "missing_percentage"
    ]
)

display(
    missing_df.orderBy(
        F.desc("missing_count")
    )
)
```

```python
%sql
--sprawdzenie unikalności student_id
SELECT
    COUNT(*) AS total_records,
    COUNT(student_id) AS non_null_ids,
    COUNT(DISTINCT student_id) AS unique_students,
    COUNT(*) - COUNT(student_id) AS null_ids,
    COUNT(student_id) - COUNT(DISTINCT student_id)
        AS duplicate_ids
FROM workspace.student_bronze.raw_student_exams;
```

```python
%sql

SELECT
    student_id,
    COUNT(*) AS record_count
FROM workspace.student_bronze.raw_student_exams
WHERE student_id IS NOT NULL
GROUP BY student_id
HAVING COUNT(*) > 1
ORDER BY record_count DESC;
```

```python
business_df = df.select(*business_columns)

total_records = business_df.count()

unique_records = (
    business_df.distinct().count()
)

print(f"Total records: {total_records:,}")
print(f"Unique records: {unique_records:,}")
print(
    f"Duplicate records: "
    f"{total_records - unique_records:,}"
)
```

```python
# unikalne wartości w kolumnach
distinct_counts = df.agg(
    *[
        F.countDistinct(c).alias(c)
        for c in business_columns
    ]
).first().asDict()

distinct_df = spark.createDataFrame(
    [
        (c, distinct_counts[c])
        for c in business_columns
    ],
    [
        "column_name",
        "distinct_values"
    ]
)

display(
    distinct_df.orderBy(
        F.asc("distinct_values")
    )
)
```

```python
#analiza zmiennych kategorycznych
categorical_columns = [
    "gender",
    "education_level",
    "school_type",
    "family_income",
    "parent_education",
    "urban_rural",
    "class_participation",
    "private_tuition",
    "study_consistency",
    "study_environment",
    "study_method",
    "revision_frequency",
    "notes_quality",
    "sleep_quality",
    "break_frequency",
    "stress_level",
    "motivation_level",
    "internet_access",
    "device_availability",
    "educational_app_usage",
    "exam_difficulty",
    "exam_anxiety_level",
    "performance_grade",
    "pass_status",
    "performance_level"
]

for column in categorical_columns:

    print(f"Column: {column}")

    display(
        df.groupBy(column)
          .count()
          .orderBy(F.desc("count"))
          .limit(30)
    )
```

```python
# analiza zmiennych numerycznych
numeric_columns = [
    "age",
    "previous_exam_score",
    "previous_gpa",
    "attendance_percentage",
    "assignment_completion_rate",
    "study_hours_per_day",
    "self_study_hours",
    "online_learning_hours",
    "practice_tests_completed",
    "sleep_hours",
    "daily_screen_time",
    "physical_activity_hours",
    "online_course_hours",
    "exam_preparation_days",
    "questions_attempted",
    "questions_correct",
    "time_management_score",
    "exam_score"
]

numeric_df = df.select(
    *[
        F.expr(
            f"TRY_CAST(`{c}` AS DOUBLE)"
        ).alias(c)
        for c in numeric_columns
    ]
)

display(
    numeric_df.summary(
        "count",
        "mean",
        "stddev",
        "min",
        "25%",
        "50%",
        "75%",
        "max"
    )
)
```

```python
# sprawdzenie błędów konwersji typów
conversion_errors = df.agg(
    *[
        F.count(
            F.when(
                F.col(c).isNotNull() &
                (F.trim(F.col(c)) != "") &
                F.expr(
                    f"TRY_CAST(`{c}` AS DOUBLE)"
                ).isNull(),
                1
            )
        ).alias(c)
        for c in numeric_columns
    ]
).first().asDict()

conversion_df = spark.createDataFrame(
    [
        (c, conversion_errors[c])
        for c in numeric_columns
    ],
    [
        "column_name",
        "conversion_errors"
    ]
)

display(
    conversion_df.orderBy(
        F.desc("conversion_errors")
    )
)
```

```python
%sql

SELECT
    COUNT_IF(
        TRY_CAST(questions_correct AS DOUBLE) >
        TRY_CAST(questions_attempted AS DOUBLE)
    ) AS invalid_questions,

    COUNT_IF(
        TRY_CAST(attendance_percentage AS DOUBLE)
        NOT BETWEEN 0 AND 100
    ) AS invalid_attendance,

    COUNT_IF(
        TRY_CAST(sleep_hours AS DOUBLE)
        NOT BETWEEN 0 AND 24
    ) AS invalid_sleep_hours,

    COUNT_IF(
        TRY_CAST(study_hours_per_day AS DOUBLE)
        NOT BETWEEN 0 AND 24
    ) AS invalid_study_hours,

    COUNT_IF(
        TRY_CAST(exam_score AS DOUBLE) < 0
    ) AS negative_exam_scores

FROM workspace.student_bronze.raw_student_exams;
```
