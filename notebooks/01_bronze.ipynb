```python
import requests
from pathlib import Path
```

```python
# źródło danych
base_url = (
    "https://raw.githubusercontent.com/"
    "Fban41/student-performance-big-data/main/data/raw/"
)

file_names = [
    "student_exam_performance.csv"
]

out_dir = Path(
    "/Volumes/workspace/student_bronze/raw_files"
)
```

```python
# pobranie danych z GitHuba
for name in file_names:

    target = out_dir / name

    if target.exists():
        print(f"Already there: {name}")
        continue

    response = requests.get(
        base_url + name,
        timeout=600
    )

    response.raise_for_status()

    target.write_bytes(response.content)

    print(
        f"Downloaded {name}: "
        f"{len(response.content):,} bytes"
    )
```

```python
%sql

-- utworzenie tabeli Bronze
CREATE OR REPLACE TABLE
    workspace.student_bronze.raw_student_exams
USING DELTA
AS

SELECT
    _metadata.file_path AS source_file,
    current_timestamp() AS loaded_at,
    *
FROM read_files(
    '/Volumes/workspace/student_bronze/raw_files/student_exam_performance.csv',
    format => 'csv',
    header => true,
    inferSchema => false
);
```
