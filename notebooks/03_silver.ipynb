```python
from pyspark.sql import functions as F
from pyspark.sql import types as T

bronze_table = "workspace.student_bronze.raw_student_exams"

silver_table = "workspace.student_silver.student_exams_clean"

rejected_table = "workspace.student_silver.student_exams_rejected"

quality_table = "workspace.student_silver.data_quality_report"

df = spark.table(bronze_table)

display(df.limit(10))
```

```python
# zdefiniowanie typów danych
integer_columns = [
    "age",
    "practice_tests_completed",
    "stress_level",
    "exam_preparation_days",
    "questions_attempted",
    "questions_correct"
]

double_columns = [
    "previous_exam_score",
    "previous_gpa",
    "attendance_percentage",
    "assignment_completion_rate",
    "study_hours_per_day",
    "self_study_hours",
    "online_learning_hours",
    "sleep_hours",
    "daily_screen_time",
    "physical_activity_hours",
    "online_course_hours",
    "time_management_score",
    "exam_anxiety_level",
    "exam_score"
]

boolean_columns = [
    "private_tuition",
    "internet_access"
]

categorical_columns = [
    "gender",
    "education_level",
    "school_type",
    "family_income",
    "parent_education",
    "urban_rural",
    "class_participation",
    "study_consistency",
    "study_environment",
    "study_method",
    "revision_frequency",
    "notes_quality",
    "sleep_quality",
    "break_frequency",
    "motivation_level",
    "device_availability",
    "educational_app_usage",
    "exam_difficulty",
    "performance_grade",
    "pass_status",
    "performance_level"
]

business_columns = [
    "student_id",
    *integer_columns,
    *double_columns,
    *boolean_columns,
    *categorical_columns
]

assert len(business_columns) == 44
assert set(business_columns).issubset(df.columns)
```

```python
# standaryzacja danych tekstowych
df_normalized = df

for c in business_columns:

    df_normalized = df_normalized.withColumn(
        c,
        F.when(
            F.trim(F.col(c)) == "",
            F.lit(None)
        ).otherwise(
            F.trim(F.col(c))
        )
    )
```

```python
# deduplikacja - teraz jest git, ale w przyszlosci jakby dane doszly to jest zabezpiecznie
rows_before = df_normalized.count()

df_deduplicated = df_normalized.dropDuplicates(
    business_columns + ["_rescued_data"]
)

rows_after = df_deduplicated.count()

duplicates_removed = rows_before - rows_after

print(f"Rows before: {rows_before:,}")
print(f"Rows after: {rows_after:,}")
print(f"Duplicates removed: {duplicates_removed:,}")
```

```python
# konwersja typów danych
df_typed = df_deduplicated

conversion_checks = []

for c in integer_columns:

    parsed = F.expr(f"TRY_CAST(`{c}` AS INT)")

    conversion_checks.append(
        F.when(
            F.col(c).isNotNull() &
            parsed.isNull(),
            F.lit(f"Invalid integer: {c}")
        )
    )

for c in double_columns:

    parsed = F.expr(f"TRY_CAST(`{c}` AS DOUBLE)")

    conversion_checks.append(
        F.when(
            F.col(c).isNotNull() &
            (
                parsed.isNull() |
                F.isnan(parsed) |
                (F.abs(parsed) == float("inf"))
            ),
            F.lit(f"Invalid number: {c}")
        )
    )

for c in boolean_columns:

    conversion_checks.append(
        F.when(
            F.col(c).isNotNull() &
            ~F.col(c).isin("0", "1"),
            F.lit(f"Invalid boolean: {c}")
        )
    )

df_typed = df_typed.withColumn(
    "conversion_errors",
    F.concat_ws("; ", *conversion_checks)
)

for c in integer_columns:

    df_typed = df_typed.withColumn(
        c,
        F.expr(f"TRY_CAST(`{c}` AS INT)")
    )

for c in double_columns:

    df_typed = df_typed.withColumn(
        c,
        F.expr(f"TRY_CAST(`{c}` AS DOUBLE)")
    )

for c in boolean_columns:

    df_typed = df_typed.withColumn(
        c,
        F.when(F.col(c) == "1", True)
         .when(F.col(c) == "0", False)
         .otherwise(F.lit(None).cast("boolean"))
    )

df_typed.printSchema()
```

```python
# obsługa brakujących wartości
missing_columns = [
    "attendance_percentage",
    "time_management_score",
    "notes_quality",
    "previous_gpa",
    "sleep_quality",
    "parent_education"
]

df_cleaned = df_typed

for c in missing_columns:

    df_cleaned = df_cleaned.withColumn(
        f"is_{c}_missing",
        F.col(c).isNull()
    )
```

```python
df_cleaned = df_cleaned.fillna(
    "Unknown",
    subset=[
        "notes_quality",
        "sleep_quality",
        "parent_education"
    ]
)
```

```python
# walidacja unikalności klucza
from pyspark.sql.window import Window

student_window = Window.partitionBy("student_id")

df_cleaned = df_cleaned.withColumn(
    "student_record_count",
    F.count("*").over(student_window)
)
```

```python
# walidacja reguł biznesowych
validation_rules = {

    "Missing student ID":
        F.col("student_id").isNull(),

    "Duplicate student ID":
        F.col("student_record_count") > 1,

    "Missing exam score":
        F.col("exam_score").isNull(),

    "Invalid exam score":
        ~F.col("exam_score").between(0, 100),

    "Invalid previous exam score":
        ~F.col("previous_exam_score").between(0, 100),

    "Invalid attendance":
        ~F.col("attendance_percentage").between(0, 100),

    "Invalid assignment completion rate":
        ~F.col("assignment_completion_rate").between(0, 100),

    "Invalid sleep hours":
        ~F.col("sleep_hours").between(0, 24),

    "Invalid study hours":
        ~F.col("study_hours_per_day").between(0, 24),

    "Invalid question count":
        (F.col("questions_correct") < 0) |
        (F.col("questions_attempted") < 0) |
        (
            F.col("questions_correct") >
            F.col("questions_attempted")
        ),

    "Invalid pass status":
        F.col("pass_status").isNull() |
        ~F.col("pass_status").isin("Pass", "Fail"),

    "Invalid performance grade":
        F.col("performance_grade").isNull() |
        ~F.col("performance_grade").isin(
            "A", "B", "C", "D", "F"
        ),

    "Rescued source data":
        F.col("_rescued_data").isNotNull(),

    "Data type conversion error":
        F.col("conversion_errors") != ""
}
```

```python
df_validated = df_cleaned.withColumn(
    "rejection_reason",
    F.concat_ws(
        "; ",
        *[
            F.when(condition, F.lit(reason))
            for reason, condition
            in validation_rules.items()
        ]
    )
)

df_validated = df_validated.withColumn(
    "is_valid",
    F.col("rejection_reason") == ""
)
```

```python
# podział na poprawne i odrzucone rekordy
df_valid = df_validated.filter(
    F.col("is_valid") == True
)

df_rejected = df_validated.filter(
    F.col("is_valid") == False
)

valid_count = df_valid.count()
rejected_count = df_rejected.count()

print(f"Valid records: {valid_count:,}")
print(f"Rejected records: {rejected_count:,}")

display(df_rejected.limit(20))
```

```python
# tworzenie nowych zmiennych; feature engineering
df_enriched = (

    df_valid

    .withColumn(
        "age_group",
        F.when(F.col("age").isNull(), "Unknown")
         .when(F.col("age") < 16, "Under 16")
         .when(F.col("age") < 18, "16-17")
         .otherwise("18+")
    )

    .withColumn(
        "study_hours_group",
        F.when(F.col("study_hours_per_day").isNull(), "Unknown")
         .when(F.col("study_hours_per_day") < 2, "Under 2h")
         .when(F.col("study_hours_per_day") < 4, "2-4h")
         .when(F.col("study_hours_per_day") < 6, "4-6h")
         .otherwise("6h+")
    )

    .withColumn(
        "sleep_hours_group",
        F.when(F.col("sleep_hours").isNull(), "Unknown")
         .when(F.col("sleep_hours") < 6, "Under 6h")
         .when(F.col("sleep_hours") < 8, "6-8h")
         .when(F.col("sleep_hours") < 10, "8-10h")
         .otherwise("10h+")
    )

    .withColumn(
        "attendance_group",
        F.when(F.col("attendance_percentage").isNull(), "Unknown")
         .when(F.col("attendance_percentage") < 70, "Under 70%")
         .when(F.col("attendance_percentage") < 80, "70-80%")
         .when(F.col("attendance_percentage") < 90, "80-90%")
         .otherwise("90-100%")
    )

    .withColumn(
        "exam_score_change",
        F.round(
            F.col("exam_score") -
            F.col("previous_exam_score"),
            2
        )
    )

    .withColumn(
        "question_accuracy_pct",
        F.when(
            F.col("questions_attempted") > 0,
            F.round(
                F.col("questions_correct") /
                F.col("questions_attempted") * 100,
                2
            )
        )
    )

    .withColumn(
        "is_passed",
        F.col("pass_status") == "Pass"
    )

    .withColumn(
        "screen_time_group",
        F.when(F.col("daily_screen_time").isNull(), "Unknown")
         .when(F.col("daily_screen_time") < 3, "Under 3h")
         .when(F.col("daily_screen_time") < 6, "3-6h")
         .otherwise("6h+")
    )
)
```

```python
# przygotowanie finalnej tabeli silver
df_silver = (

    df_enriched

    .drop(
        "_rescued_data",
        "conversion_errors",
        "student_record_count",
        "rejection_reason",
        "is_valid"
    )

    .withColumn(
        "processed_at",
        F.current_timestamp()
    )
)

display(df_silver.limit(10))
```

```python
# zapis tabeli clean
(
    df_silver.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(silver_table)
)
```

```python
# zapis tabeli rejected
(
    df_rejected

    .withColumn(
        "rejected_at",
        F.current_timestamp()
    )

    .write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(rejected_table)
)
```

```python
# jakosc danych
quality_metrics = [

    ("bronze_row_count", rows_before),

    ("duplicate_rows_removed", duplicates_removed),

    ("silver_valid_rows", valid_count),

    ("silver_rejected_rows", rejected_count)
]

missing_counts = df_normalized.agg(
    *[
        F.count(
            F.when(F.col(c).isNull(), 1)
        ).alias(c)
        for c in business_columns
    ]
).first().asDict()

for c in business_columns:

    quality_metrics.append(
        (f"missing_{c}", missing_counts[c])
    )

quality_df = spark.createDataFrame(
    quality_metrics,
    ["metric_name", "metric_value"]
)

quality_df = (

    quality_df

    .withColumn(
        "checked_at",
        F.current_timestamp()
    )

    .withColumn(
        "source_table",
        F.lit(bronze_table)
    )
)

display(quality_df)
```

```python
# zapis raportu o jakości danych
(
    quality_df.write
    .format("delta")
    .mode("append")
    .saveAsTable(quality_table)
)
```
